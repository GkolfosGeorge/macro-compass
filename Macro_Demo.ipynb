# Macro Regime Analyzer — Demo

End-to-end walkthrough of the Macro Regime Analyzer.

**Flow:** Load data → Current Snapshot → Interactive Charts → Print-Ready Charts → Historical Analysis

> Data is downloaded automatically via `yfinance`. First run may take ~30 seconds. Subsequent runs load from local cache instantly.

## 1. Imports

```python
import sys, os
import plotly.io as pio
pio.renderers.default = "notebook"


import pandas as pd
import matplotlib.pyplot as plt

from macro_analyzer import MacroAnalyzer
from macro_report   import print_macro_snapshot, print_crisis_analysis
from macro_charts   import phase_timeline, probit_recession, crisis_heatmap, cg_yield_divergence, spy_sma200
from macro_charts_print import (
    phase_gantt_print,
    ratios_history_print,
    crisis_heatmap_print,
    sector_rotation_wheel_print,
    probit_recession_print,
)
```

## 2. Load Data

Downloads macro data from 2000 via `yfinance`. Pass a FRED API key to unlock additional signals (Real Yields, TED Spread, SOFR).

```python
ma = MacroAnalyzer(
    # fred_api_key = "your_key_here"  # optional — get one free at fred.stlouisfed.org
)

ma.load(start="2000-01-01")
```

## 3. Current Macro Snapshot

Classifies the current environment into a cycle phase and outputs all signal readings.

```python
snap = ma.get_snapshot()
print_macro_snapshot(snap)
```

## 4. Interactive Charts (Plotly)

Five core charts covering regime history, recession probability, crisis signals, macro divergence, and equity regime.

```python
# Cycle phase timeline — full history from 2000
phase_timeline(ma).show()
```

```python
# Probit recession probability (Estrella & Mishkin model)
probit_recession(ma).show()
```

```python
# Pre-crisis signal heatmap — how signals behaved before each historical crisis
crisis_heatmap(ma).show()
```

```python
# Copper/Gold vs 10Y yield divergence — leading macro signal
cg_yield_divergence(ma).show()
```

```python
# SPY vs SMA200 — bull/bear equity regime
spy_sma200(ma).show()
```

## 5. Print-Ready Charts (Matplotlib)

Static 300dpi charts — suitable for reports, presentations, and PDF export.

```python
# 04a — Cycle phase Gantt chart (full history overview)
fig = phase_gantt_print(ma)
plt.show()
```

```python
# 06 — Commodity ratios history (Copper/Gold, Gold/Silver, Gold/Oil)
fig = ratios_history_print(ma)
plt.show()
```

```python
# 10 — Pre-crisis signal heatmap (print version)
fig = crisis_heatmap_print(ma)
plt.show()
```

```python
# 13 — Sector rotation wheel by cycle phase
fig = sector_rotation_wheel_print(snap)
plt.show()
```

```python
# 16 — Probit recession probability (print version)
fig = probit_recession_print(ma)
plt.show()
```
